```python
import numpy as np
import matplotlib.pyplot as plt 

def f(x):
    return (-x**5 + 4*x**4 - 4*x**3 + x**2*np.exp(x) - 2*x**2 - 4*x*np.exp(x) + 8*x + 4*np.exp(x) - 8)

def df(x):
    return (-5*x**4 + 16*x**3 - 12*x**2 + (x**2 + 2*x)*np.exp(x) - 4*x - 4*(x + 1)*np.exp(x) + 8 + 4*np.exp(x))

# Setting up tolerance and maximum iterations
tol = 1.e-8
kmax = 200

# FIXED-POINT
def fixed_point(f, x0, a=1):
    xs = [x0]
    x = x0
    for k in range(1, kmax):
        xnew = x + a*f(x)
        xs.append(xnew)
        if abs(xnew - x) < tol:
            break
        x = xnew 
    else:
        print("fixedpoint: maxed out")
    return np.array(xs) 

# Setting initial conditions 
x0 = -1.5
a = 0.01
xs = fixed_point(f, x0, a)
root = xs[-1]
print(f"root = {root:.8f}, steps = {len(xs)-1}")

# Plotting FIXED-POINT
iterations = np.arange(len(xs))

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(iterations, xs, 'o-', color='navy',
        label=f'Fixed-Point ($x_0$ = {x0}, $a$ = {a})')
ax.axhline(root, color='gray', linestyle='--', label=f'Root = {root:.4f}')

ax.set_xlabel('Iteration step ($i$)')
ax.set_ylabel('$x_i$')
ax.set_title('Convergence of Fixed-Point Method ($x_i$ vs $i$)')

ax.grid(False)
ax.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt 

def f(x):
    return (-x**5 + 4*x**4 - 4*x**3 + x**2*np.exp(x) - 2*x**2 - 4*x*np.exp(x) + 8*x + 4*np.exp(x) - 8)

def df(x):
    return (-5*x**4 + 16*x**3 - 12*x**2 + (x**2 + 2*x)*np.exp(x) - 4*x - 4*(x + 1)*np.exp(x) + 8 + 4*np.exp(x))

# Setting up tolerance and maximum iterations
tol = 1.e-8
kmax = 200

# BISECTION
def bisection(f, x0, x1):
    if f(x0)*f(x1) > 0:
       raise ValueError("bisection: no sign change in bracket")
    xs = [] 
    for k in range(1, kmax):
        xs_k = 0.5*(x0 + x1)
        xs.append(xs_k)
        if f(x0)*f(xs_k) < 0:
            x1 = xs_k
        else:
            x0 = xs_k
        if abs(x1 - x0) < tol:
            break 
    else:
        print("bisection: maxed out")
    return np.array(xs) 

# Setting initial conditions 
x0 = -2
x1 = 0
a = 0
xs = bisection(f, x0, x1)
root = xs[-1]
print(f"root = {root:.8f}, steps = {len(xs)-1}")

# Plotting FIXED-POINT
iterations = np.arange(len(xs))

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(iterations, xs, 'o-', color='navy',
        label=f'Bisection (bracket = [{x0}, {x1}])')
ax.axhline(root, color='gray', linestyle='--', label=f'Root = {root:.4f}')

ax.set_xlabel('Iteration step ($i$)')
ax.set_ylabel('$x_i$')
ax.set_title('Convergence of Bisection Method ($x_i$ vs $i$)')

ax.grid(False)
ax.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt 

def f(x):
    return (-x**5 + 4*x**4 - 4*x**3 + x**2*np.exp(x) - 2*x**2 - 4*x*np.exp(x) + 8*x + 4*np.exp(x) - 8)

def df(x):
    return (-5*x**4 + 16*x**3 - 12*x**2 + (x**2 + 2*x)*np.exp(x) - 4*x - 4*(x + 1)*np.exp(x) + 8 + 4*np.exp(x))

# Setting up tolerance and maximum iterations
tol = 1.e-8
kmax = 200

# NEWTON
def newton(f, df, x0):
    xs = [x0]
    x = x0
    for k in range(1, kmax):
        xnew = x - f(x)/df(x)
        xs.append(xnew)
        if abs(xnew -x) < tol:
            break
        x = xnew 
    else:
        print("newton: maxed out")
    return np.array(xs)
    
# Setting initial conditions 
x0 = -2
xs = newton(f, df, x0)
root = xs[-1]
print(f"root = {root:.8f}, steps = {len(xs)-1}")

# Plotting Newton
iterations = np.arange(len(xs))

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(iterations, xs, 'o-', color='navy',
        label=f'Newton ($x_0$ = {x0})')
ax.axhline(root, color='gray', linestyle='--', label=f'Root = {root:.4f}')

ax.set_xlabel('Iteration step ($i$)')
ax.set_ylabel('$x_i$')
ax.set_title('Convergence of Newton Method ($x_i$ vs $i$)')

ax.grid(False)
ax.legend()

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt 

def f(x):
    return (-x**5 + 4*x**4 - 4*x**3 + x**2*np.exp(x) - 2*x**2 - 4*x*np.exp(x) + 8*x + 4*np.exp(x) - 8)

def df(x):
    return (-5*x**4 + 16*x**3 - 12*x**2 + (x**2 + 2*x)*np.exp(x) - 4*x - 4*(x + 1)*np.exp(x) + 8 + 4*np.exp(x))

# Setting up tolerance and maximum iterations
tol = 1.e-8
kmax = 200

# SECANT
def secant(f, x0, x1):
    xs = [x0, x1]
    for k in range(1, kmax):
        xnew = x1 - f(x1)*(x1-x0) / (f(x1) - f(x0))
        xs.append(xnew)
        if abs(xnew -x1) < tol:
            break
        x0, x1 = x1, xnew
    else:
        print("secant: maxed out")
    return np.array(xs)
    
# Setting initial conditions 
x0 = -2
x1 = -1.5
xs = secant(f, x0, x1)
root = xs[-1]
print(f"root = {root:.8f}, steps = {len(xs)-1}")

# Plotting Secant
iterations = np.arange(len(xs))

fig, ax = plt.subplots(figsize=(8,5))

ax.plot(iterations, xs, 'o-', color='navy',
        label=f'Secant ($x_0$ = {x0}, $x_1$ = {x1})')
ax.axhline(root, color='gray', linestyle='--', label=f'Root = {root:.4f}')

ax.set_xlabel('Iteration step ($i$)')
ax.set_ylabel('$x_i$')
ax.set_title('Convergence of Secant Method ($x_i$ vs $i$)')

ax.grid(False)
ax.legend() 

plt.tight_layout()
plt.show()
```
