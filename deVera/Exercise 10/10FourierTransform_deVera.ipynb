```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(-10, 10, 1000)
dt = t[1] - t[0]

freqs = np.fft.fftfreq(len(t), dt)
freqs_shifted = np.fft.fftshift(freqs)

fig, axs = plt.subplots(1, 3, figsize=(12, 4))

sigmas = [0.5, 1.0, 2.0]

for s in sigmas:
    # signal
    y = np.exp(-(t**2) / (2 * s**2))
    
    # fourier
    yf = np.fft.fft(y)
    yf_shifted = np.fft.fftshift(yf)
    
    # magnitude
    mag = np.abs(yf_shifted) * dt
    
    # phase
    phase = np.angle(yf_shifted)
    
    phase[mag < 0.00001] = 0 
    
    # plotting
    axs[0].plot(t, y, label=f'sigma = {s}')
    axs[1].plot(freqs_shifted, mag, label=f'sigma = {s}')
    axs[2].plot(freqs_shifted, phase, label=f'sigma = {s}')

# signal plot
axs[0].set_title('Gaussian Signal')
axs[0].set_xlim(-5, 5)
axs[0].legend()

# magnitude plot
axs[1].set_title('Magnitude')
axs[1].set_xlim(-3, 3)
axs[1].legend()

# phase plot
axs[2].set_title('Phase')
axs[2].set_xlim(-3, 3)
axs[2].legend()

plt.tight_layout()
plt.show()
```
