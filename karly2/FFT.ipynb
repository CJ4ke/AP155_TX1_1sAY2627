```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
```

```python
N_sample = 1000
# N_unaliased = 1000
t = 10
dt = t/N_sample
# dt_un = t/N_unaliased
t_sample = np.arange(N_sample)*dt
# t_un = np.arange(N_unaliased)*dt_un
# signal = lambda t, freq: np.cos(2*np.pi*freq*t)
#sq_signal = lambda t, freq: signal.square(2*np.pi*freq*t)
dd_signal = lambda freq: signal.unit_impulse(int(1000), N_sample//2)
```

```python
freq = 7.5
disc_signal = dd_signal(freq)
# disc_signal_un= signal(t_un, freq)
```

```python
fig, ax = plt.subplots(figsize = (12,5))

ax.plot(t_sample, disc_signal)
# ax.plot(t_un, disc_signal_un)
fig.show()
```

```python
freqs = np.fft.fftfreq(N_sample, dt)
freqs = np.fft.fftshift(freqs)
```

```python
Yk1 = np.fft.fftshift(np.fft.fft(disc_signal))
Yk_mag1 = np.abs(Yk1)
Yk_phase1 = np.unwrap(np.angle(Yk1))
```

```python
plt.plot(freqs, Yk_mag1)
```
