```python
N = 1000
# function endpoints (f1:x1, f2:x2, ...)
x1 = np.linspace(0, 3/2, N) 
x2 = np.linspace(-2, 2, N)
x3 = np.linspace(0, np.pi, N)

f1 = lambda x: np.sqrt(9 - 4*x**2)
f2 = lambda x: -x**2 + 1
f3 = lambda x: np.cos(x)*(np.sqrt(x))

#functions dictionary 
functions= { f1: {"label": r"$f_1(x) = \sqrt{9-4x^2}$", "a": 0, "b": 1.5},
             f2: {"label": r"$f_2(x) = -x^2 + 1$", "a": 0, "b": 1.0},
             f3: {"label": r"$f_3(x) = \cos(x)\sqrt{x}$", "a": 0, "b": np.pi}}
```

```python
# Trapezoid

def trapezoid(f, a, b, n):
    h = (b-a)/(n-1)
    xs = a + np.arange(n)*h
    cs = np.ones(n); cs[0] = 0.5; cs[-1]=0.5
    contribs = cs*f(xs)
   
    signed_area_trap = h*np.sum(contribs)
    total_area_trap = h*np.sum(np.abs(contribs))

    return signed_area_trap, total_area_trap

print(trapezoid(f1, 0, 3/2, 51))
```

```python
# Simpson 1/3

def simpson_13(f, a, b, n):
    if n%2 == 0: #constraint: odd n
        n += 1
    h = (b-a)/(n-1)
    xs = a + np.arange(n)*h
    cs = 2*np.ones(n)
    cs[1::2] = 4; cs[0] = 1; cs[-1] = 1 # 1,4,2,4,2,...,2,4,1
    contribs = cs*f(xs)
    
    signed_area_13 = (h/3)*np.sum(contribs)
    total_area_13 = (h/3)*np.sum(np.abs(contribs))

    return signed_area_13, total_area_13

print(simpson_13(f1, 0, 3/2, 51))
```

```python
# Simpson 3/8
def simpson_38(f, a, b, n):
    if (n-1)%3 != 0:
        n = n + (3-(n-1) % 3) #constraint: needs 3n+1 points)
    h = (b-a)/(n-1)
    xs = a + np.arange(n)*h
    cs = 3*np.ones(n)
    cs[0::3] = 2; cs[0] = 1; cs[-1] = 1 # 1,3,3,2,3,3,...,2,3,3,1
    contribs = cs*f(xs)
    
    signed_area_38 = (h*(3/8))*np.sum(contribs)
    total_area_38 = (h*(3/8))*np.sum(np.abs(contribs))

    return signed_area_38, total_area_38
print(simpson_38(f1, 0, 3/2, 51))
```

```python
# Error plot
def plot_error(f, a, b, label, n=10000):
    x = np.linspace(a,b,n)
    method = {'Trapezoid': trapezoid, 'Simpsons 1/3': simpson_13, 'Simpsons 3/8': simpson_38} #methods dictionary
    
    analytic_signed, _ = quad(f, a, b) #since it has two outputs

    n_values = np.unique(np.geomspace(5,1000,50, dtype=int)) #generating different log grid sizes and ensuring that all values are integers

    plt.figure(figsize = (8,4))
    for method_name, method_func in method.items():
        errors = []
        plot_n = []
        for n in n_values:
            actual_n = n
            if method_name == 'Simpsons 1/3' and actual_n % 2 == 0: #constraints in n
                actual_n += 1
            elif method_name == 'Simpsons 3/8' and (actual_n - 1) % 3 != 0: #constraints in n
                actual_n = actual_n + (3 - (actual_n - 1) % 3)
            comp_signed, _ = method_func(f,a,b,n)
            error = abs(comp_signed - analytic_signed)
            errors.append(error)
            plot_n.append(actual_n)
        plt.loglog(plot_n, errors, 'o-', label=method_name, markersize = 4)

    plt.xlabel('Number of Points ($n$)', fontsize=9)
    plt.ylabel('Absolute Error', fontsize=9)
    plt.title(f'Convergence for {label} ', fontsize=12)
    plt.grid(which='both', linestyle=":", alpha=0.7)
    plt.legend(fontsize=10)
    plt.tight_layout()
    plt.show()

for func, info in functions.items():
    plot_error(func, info["a"], info["b"], info["label"])
```

```python
# Monte Carlo

def montecarlo(f, a, b, n, option = "uniform"):
  np.random.seed(42067)
  us = np.random.uniform(a, b, n)
  if option == "uniform":
    fs = f(us)
  else:
    c0 = 4 - np.sqrt(2)
    c1 = -6 + 4*np.sqrt(2)
    xs = (-c0 + np.sqrt(2*c1*us + c0**2))/c1
    fs = f(xs)/(c0+c1*xs)

  fbar, err = stats(fs)
  return (b-a)*fbar, (b-a)*err

def stats(fs):
  n = fs.size
  fbar = np.sum(fs)/n
  fsq = np.sum(fs**2)/n
  varfbar = (fsq-fbar**2)/(n-1)
  return fbar, np.sqrt(varfbar)

for n in 10**np.arange(2,7):
  avu, erru = montecarlo(f1, 0, 3/2, n)
  avi, erri = montecarlo(f1, 0, 3/2, n, option="is")
  rowf = "{0:7d} {1:1.9f} {2:1.9f} {3:1.9f} {4:1.9f}"
  print(rowf.format(n, avu, erru, avi, erri))
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import quad

def plot_montecarlo(ax, f, a, b, n, name="Function"):
    signed_area, error = montecarlo(f, a, b, n, option="uniform")
    analytic_int, _ = quad(f, a, b)
    
    x_m = np.linspace(a, b, 500)
    y_m = f(x_m)
    ymax = max(0, np.max(y_m))
    ymin = min(0, np.min(y_m))

    # Generating random points
    xs = np.random.uniform(a, b, n)
    ys = np.random.uniform(ymin, ymax, n)
    
    # Regions
    above = (ys > 0) & (ys <= f(xs))  # Above x-axis (blue)
    below = (ys < 0) & (ys >= f(xs))  # Below x-axis (red)
    outside = ~(above | below)        # Not bounded above or below

    ax.scatter(xs[outside], ys[outside], color='gray', label=r'Outside $f$')
    if np.any(above):
        ax.scatter(xs[above], ys[above], color='blue', label=r'Above $x$-axis')
    if np.any(below):
        ax.scatter(xs[below], ys[below], color='red', label=r'Below $x$-axis')

    ax.plot(x_m, y_m, color='coral', linewidth=2, label=r'$f(x)$')
    ax.axhline(0, color='black', linestyle='--', lw=1)
    ax.axvline(a, color='black', linestyle='--', lw=0.7)
    ax.axvline(b, color='black', linestyle='--', lw=0.7)

    ax.set_title(f"{name} ($n={n}$)", fontsize=11)
    ax.set_xlabel('$x$')
    ax.set_ylabel('$y$')
    ax.grid(linestyle=':', alpha=0.3)
    ax.legend()

    print(f"--- {name} ---")
    print(f"    Signed Area: {signed_area:.6f}")
    print(f"    Error: {error:.6f}")
    print(f"    Analytic Integral: {analytic_int:.6f}\n")


fig, axes = plt.subplots(2, 2, figsize=(12, 8))

# Assuming functions format: {"Label": (func, (a, b))}
for ax, (name, (f, (a, b))) in zip(axes.flat, functions.items()):
    plot_montecarlo(ax, f, a, b, 3000, name=name)

# Remove unused 4th subplot
fig.delaxes(axes[1, 1])

plt.tight_layout()
plt.show()
```
