### Problem 1:

Consider the function:
$$ f(x) = 2 + 5x\sin\left(x + \frac{\pi}{4}\right)$$

Say you start with 257 points from 0 to $4\pi$ 

```python
f_0 = lambda x: 2 + 5*x*np.sin(1*x+np.pi/4)
x = np.linspace(0, 4*np.pi, 257)
fig, ax = plt.subplots(figsize = (7,5))
ax.scatter(x, f_0(x))
ax.set_ylabel(r"$2 + 5x\sin\left(x + \frac{\pi}{4}\right)$")
ax.set_xlabel("x")
fig.show()
fig.savefig('f_x.png')
```

- Create a function that gets the first derivative approximation using the forward, backward, and central difference formulas. (Bonus, not required): Make the function flexible to non-evenly spaced data
- To give yourself a visualization, plot (in a single axis):
  - the three derivative approximations
  - the analytically derived function

- Get the mean absolute error of each of the approximations compared to the analytically derived function
- Repeat the above step for increasing number of points (decreasing spacing) - instead of 256 points, take 512/1024/128/64 points, etc.

- (For the report) Make a plot of the mean absolute error vs the spacing between points. In this plot, there should be three curves: One for the forward, one for the backward, and one for the central difference approximation

```python
def forward_difference(x, t):
    ''' 
    given: evenly spaced numpy array x
    '''

    return (x - np.roll(x,1))[1:]/(t[1]-t[0]), t[1:]

def backward_difference(x,t):
    ''' 
    given: evenly spaced numpy array x
    '''

    return (np.roll(x,-1)-x)[:-1]/(t[1]-t[0]), t[1:]


def central_difference(x,t):
    ''' 
    given: evenly spaced numpy array x
    '''

    return (np.roll(x,-1) - np.roll(x,1))[1:-1]/(2*(t[1]-t[0])), t[1:-1]
    
    
```

```python
start = 0
end = 4*np.pi
num_points = 256+1

t = np.linspace(start,end,num_points)
f = lambda x: 2 + 5*x*np.sin(1*x+np.pi/4) #analytical function
f1prime_anafunc = lambda x: (5*np.sin(x+np.pi/4) + 5*x*np.cos(x+np.pi/4)) #analytical derivative of the function
```

```python


f1prime_forward = forward_difference(f(t),t)
f1prime_backward = backward_difference(f(t),t)
f1prime_central = central_difference(f(t),t)
```

```python
# For visualizing only
fig, ax = plt.subplots(figsize = (7,5))
ax.scatter(t, f1prime_anafunc(t), alpha = 0.3, label = 'analytical')
ax.scatter(f1prime_forward[-1], f1prime_forward[0], alpha = 0.3, label = 'forward')
ax.scatter(f1prime_backward[-1], f1prime_backward[0], alpha = 0.3, label = 'backward')
ax.scatter(f1prime_central[-1], f1prime_central[0], alpha = 0.3, label = 'central')
ax.legend()
ax.set_xlabel('t')
ax.set_ylabel(r"$\frac{df}{dt}$")
fig.show()
```

```python
#Getting the mean absolute error

def gettingMAE(approx_res, f_ana, t):
    ana_res = f_ana(t)
    #print(approx_res, ana_res)
    error = approx_res - ana_res 
    return np.mean(abs(error))
    
```

```python
#varying the step size 
num_points_list = np.array([2**x for x in np.arange(6,22,0.25)], dtype = np.int64) + 1

diff_funcs = {"forward": forward_difference, "backward":backward_difference, "central": central_difference}
num_points_list
```

```python
def calculateMAE_fromDifferentiation(f_ana, fprime_ana, diff_funcs, start, end, num_points_list):
    mae = dict()
    for diffApproach in diff_funcs:
        mae[diffApproach] = [0]*len(num_points_list)
        for i, num_points in enumerate(num_points_list):
            t_0 = np.linspace(start, end, num_points, dtype=np.float64)
            approx_res, t = diff_funcs[diffApproach](f_ana(t_0),t_0)
            mae_ = gettingMAE(approx_res, fprime_ana, t)
            mae[diffApproach][i] = mae_
    return mae

```

```python
mae = calculateMAE_fromDifferentiation(f, f1prime_anafunc, diff_funcs, start, end, num_points_list)
spacing = (end - start)/(num_points_list - 1)
```

```python
spacing
```

```python
fig, ax = plt.subplots(figsize = (7,5))
for method in mae:
    ax.scatter(spacing, mae[method], label = method, alpha = 0.4)
ax.legend()
ax.set_ylabel('mean absolute error')
ax.set_xlabel('step size')
ax.set_xscale('log')
ax.set_yscale('log')
fig.show()
```

## Section 3 (Notes)

### Non-central difference formulas (First Derivative)

* Forward 
$$
f'(x_i) = \frac{f(x_{i+1}) - f(x_{x})}{h} + O(h)
$$

* Backward
$$
f'(x_i) = \frac{f(x_{i}) - f(x_{i-1})}{h} + O(h)
$$



### Central-difference formulas

* First Derivative
$$
f'(x_i) = \frac{f(x_{i+1}) - f(x_{i-1})}{2h} + O(h^2)
$$

* Second Derivative
$$
f''(x_i) = \frac{f(x_{i+1}) - 2f(x_{i}) + f(x_{i-1})}{h^2} + O(h^2)
$$

Note: to avoid confusion, think: $x_i$ is `x[i]`, and $h$ is the gap between `x[i+1]` and `x[i]`.
