```python
def random_integer_matrix(nrows, ncols):
    matrix = []
    for i in range(nrows):
        row = []
        for i in range(ncols):
            row.append(random.randint(1, 9))
        matrix.append(row)
    return matrix

random_integer_matrix(3, 4)
```

```python
def scalar_product(list1, list2):
    elements_sum = 0
    for i in range(len(list1)):
        elements_product = list1[i] * list2[i]
        elements_sum += elements_product
    return elements_sum

sample_list1 = [2, 3, 7, 8]
sample_list2 = [3, 1, 3, 5]

scalar_product(sample_list1, sample_list2)
```

```python
def transpose_matrix(orig_matrix):
    nrows = len(orig_matrix)
    ncols = len(orig_matrix[0])
    transposed_matrix = []
    for k in range(ncols):
        new_row = []
        for i in range(nrows):
            new_row.append(orig_matrix[i][k])       
        transposed_matrix.append(new_row)
    return transposed_matrix

sample_matrix1 = [[2, 3, 7, 8], [3, 1, 3, 5], [5, 5, 1, 4]]

transpose_matrix(sample_matrix1)
```

```python
def matrix_multiply(left_matrix, right_matrix):
    if len(left_matrix[0]) == len(right_matrix): #matrix multiplication is valid
        transposed_right_matrix = transpose_matrix(right_matrix)
        product_matrix = []
        for i in range(len(left_matrix)):
            new_row = []
            for k in range(len(transposed_right_matrix)):
                new_row.append(scalar_product(left_matrix[i], transposed_right_matrix[k]))
            product_matrix.append(new_row)
        return product_matrix

    else:
        print("Matrix multiplication is not possible.")
```
