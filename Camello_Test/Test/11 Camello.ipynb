```python
import numpy as np
import matplotlib.pyplot as plt

def f1(x):
    return np.sqrt(9 - (4 * (x ** 2)))

def f2(x):
    return -x**2 + 1

def f3(x):
    return np.cos(x) * np.sqrt(x)

def rectangle(func, left, right, n):
    x_rect = np.linspace(left, right, n)
    rect_func = func(x_rect)
    rect_dx = rect_func[1] - rect_func[0]

    area = 0
    for i in rect_func:
        current_area = rect_func[i+1] * rect_dx
        area += current_area

```
