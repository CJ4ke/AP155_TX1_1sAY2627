### Problem 1:

This problem discussed error buildup when trying to evaluate polynomials without and with the use of *Horner's Rule*. Take a polynomial of degree $n-1$:
$$ P(x) = p_0 + p_1x + p_2x^2 + p_3x^3 + ... + p_{n-1}x^{n-1}. $$

Write a function that takes in a list containing the $p_i$'s, say `coeffs`, and the point $x$ and evaluates the value of $P(x)$ in the naive way, i.e. from left to right.

Notice that this way of coding up the polynomial contains several (needless) multiplications. This is so because $x^i$ is evaluated as $x\times x\times ... \times x$ (where there are $i-1$ multiplications). Thus, this way of approaching the problem corresponds to
$$ 1 + 2 + ... + n-2 = \frac{(n-1)(n-2)}{2}$$
multiplications, from $x^2$ all the way up to $x^{n-1}$. 

Let `coeffs=[(-11)**i for i in reversed(range(8))]` and `x = 11.01`

```python
coeffs= [(-11)**i for i in reversed(range(8))]
x = 11.01

print(coeffs)
```

```python
def eval_naive(coeffsList, x):
    """evaluate naively from left to right, using the power/exponent operations"""
    P = 0
    for i, p in enumerate(coeffsList):
        # print(i, p) - for my assurance
        P += p * (x ** i) # p_i * x^i
    return P

print(f"Naive Evaluation: {eval_naive(coeffs, x)}")
```

If we write the polynomial: 
$$ P(x) = p_0 + x(p_1 + x(p_2 + x(p_3 + ... + x(p_{n-2} + xp_{n-1})...))) \tag{1}$$

then we can get away with only $n-1$ multiplications (i.e., no powers are evaluated). This is obviously more efficient, but equally important is the fact that this way we can substantially limit the accumulation of rounding error (especially for polynomials of large degree).

Write a function that takes in a list containing the $p_i$'s, say `coeffs`  and the point $x$ and evaluates the value of $P(x)$ in a new way, i.e., from right to left.

```python
# this is my sanity check
reverse = list(reversed(coeffs))
print(reverse) 
```

```python
def eval_horner(coeffslist, x):
    """using Horner's rule from right to left"""
    coeffslist = list(reversed(coeffslist)) # we reverse the list so we dont use negative numbers
    val = coeffslist[0]
    for coeff in coeffslist[1:]: # we start from 1
        val = coeff + x*val
    return val

print(f"Horner's Evaluation: {eval_horner(coeffs, x)}")
```

Apply the previous two functions to the (admittedly artificial) case of:
`coeffs=[(-11)**i for i in reversed(range(8))]` and $x=11.01$. Observe any discrepancy and discuss its origin.

```python
naive_val = eval_naive(coeffs, x)
horner_val = eval_horner(coeffs, x)

print(f"Naive evaluation:   {naive_val:.16f}")
print(f"Horner's evaluation: {horner_val:.16f}")

error = 100 * abs(naive_val - horner_val) / max(naive_val, horner_val)
print(f'Error: {error}%') # use absolute error over max value 
```

We can see that there is an error of $5.24 \times 10^{-12}\%$ when comparing the two different evaluation methods. This arises as we have to perform more multiplications when naively calculating. As such, round-off errors accumulate as we evaluate. Meanwhile, using Horner's rule, we only multiply $n-1$ times, reducing the amount of round-off errors.

As we can see, naive evaluation of polynomials take significantly longer than Horner's evaluation. We can show this further by graphing the runtime graphs across $x$ values and across various degrees of polynomials

```python
import time

interval = 0.01
x_vals = np.arange(0, 1000 + interval, interval)

naive_time = [] # we put runtime here
horner_time = []

runtime_naive = 0
runtime_horner = 0

for x in x_vals:
    # naive runtime evaluation
    start_time = time.perf_counter()
    eval_naive(coeffs, x)
    end_time = time.perf_counter()
    runtime_naive += end_time - start_time
    naive_time.append(runtime_naive)
    
    # horner runtime evaluation
    start_time = time.perf_counter()
    eval_horner(coeffs, x)
    end_time = time.perf_counter()
    runtime_horner += end_time - start_time
    horner_time.append(runtime_horner)

# plot plot
fig, ax = plt.subplots(1, 1, figsize=(6, 4.5))

ax.plot(x_vals, naive_time, label = "Naive Evaluation", color = "b")
ax.plot(x_vals, horner_time, label = "Horner's Evaluation", color = "r")
ax.legend()
ax.set_title("Runtime of Evaluation Methods")
ax.set_xlabel(r"$x$", fontsize=14)
ax.set_ylabel("Cumulative Time (s)", fontsize=14)

plt.tight_layout()
plt.savefig("Figures/runtime_x.png")
plt.show()
```

```python
import time

interval = 1
degree = np.arange(1, 250, interval)
x = 11.01

naive_time = [] # we put runtime here
horner_time = []

runtime_naive = 0
runtime_horner = 0
resolution = 1000 # how many data points to average

for n in degree:
    coeffs = list(np.random.rand(n + 1)) # generate random coefficients for each polynomial degree
    
    # naive runtime evaluation
    start_time = time.perf_counter()
    for _ in range(resolution): 
        eval_naive(coeffs, x)
    end_time = time.perf_counter()
    naive_time.append((end_time - start_time)/resolution)
    
    # horner runtime evaluation
    start_time = time.perf_counter()
    for _ in range(resolution):
        eval_horner(coeffs, x)
    end_time = time.perf_counter()
    runtime_horner = end_time - start_time
    horner_time.append((end_time - start_time)/resolution)

# plot plot
fig, ax = plt.subplots(1, 1, figsize=(6, 4.5))

ax.plot(degree, naive_time, label = "Naive Evaluation", color = "b")
ax.plot(degree, horner_time, label = "Horner's Evaluation", color = "r")
ax.legend()
ax.set_title("Runtime of Evaluation Methods (varying dergee)")
ax.set_xlabel(r"Polynomial Degree $(n)$", fontsize=14)
ax.set_ylabel("Runtime (s)", fontsize=14)

plt.tight_layout()
plt.savefig("Figures/runtime_degree.png")
plt.show()
```

As expected, naively multiplying polynomials results in a higher runtime. If we investigate the number of operations needed to compute the function, using Big O notation, we find it has $O(n^2)$ operations, i.e. the runtime scales quadratically as the degree $n$ increases. On the other hand, using Horner's rule, the runtime now scales linearly ($O(n)$).

(b) Create a new Python function that codes up the following expression:
$$ s(x)= 4− \frac{3(x−2)[(x−5)^2 + 4]}{x + (x−2)^2[(x−5)^2 + 3]}$$
which is a rewritten version of our rational function. Apply this new function for
the previous two plots and compare the rounding error pattern, size, etc.

```python
def s(x):
    num = 3 * (x - 2) * ((x - 5)**2 + 4)
    den = x + (x - 2)**2 * ((x - 5)**2 + 3)
    return 4 - num/den
```

```python
# using x1 and x2 from earlier
y1 = s(x1)
y2 = s(x2)

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))

ax1.scatter(i, y1, s=1, c='k')
ax1.set_title("Algebraically equivalent $r(x_1)$")
ax1.set_xlabel(r'$1.606 + 2^{-52}i$', fontsize=14)
ax1.set_ylabel(r'$r(x)$', fontsize=14)
ax1.set_ylim(bottom=min(y1), top=max(y1))

ax2.scatter(i, y2, s=1, c='k')
ax2.set_title("Algebraically equivalent $r(x_2)$")
ax2.set_xlabel(r'$2.400 + 2^{-52}i$', fontsize=14)
ax2.set_ylabel(r'$r(x)$', fontsize=14)

plt.tight_layout()
```

In the graph above, there are fewer errors and the function is more precise as seen by the tightly packed plots forming a line. This is because we eliminate the occurence of catastrophic cancellation.

Catastrophic cancellation occurs for Horner's rule as we are subtracting large almost equal numbers. When we rewrite, we reduce the occurence of this by lowering the number of operations used in the function. Notably, in $[(x-5)^2 + 4]$ and $[(x-5)^2 + 3]$, the $(x-5)$ term is squared and thus, is always positive.
