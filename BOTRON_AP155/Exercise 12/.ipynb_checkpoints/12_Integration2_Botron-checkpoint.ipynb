### Report discussion

Figure 1 illustrates the spatial intensity distribution of a point light source viewed through a telescope. The intensely bright central region corresponds to the $r \to 0$ limit, where constructive interference is absolute. Because this central peak contains essentially most of the optical power, a raw visual evaluation obscures the surrounding behavior. For the reporting of the figure, I applyied a visual threshold (vmax=0.01) to cap the central brightness. This allows the physically faint outer maxima (the surrounding concentric rings) to become clearly visible on the screen. The dark annular nodes separating these bright rings represent regions of complete destructive interference. Mathematically, these dark regions correspond directly to the spatial roots of the governing Bessel function $J_1(kr)$ (see part A of problem 1), showing the function's oscillating pattern.  

In the previous exercise, the standard Monte Carlo method evaluated integrals by sampling uniformly across the domain. This brute-force approach is computationally inefficient for the Gamma function, as the integrand $t^{z-1}e^{-t}$ peaks sharply before rapidly decaying, which means that most samples do not really contribute to the sum. Importance sampling solves this by drawing samples from a specific probability distribution, $p(t) = e^{-t}$, which closely mirrors the target integrand's shape. This significantly reduces the variance and computational cost. As observed in Figure 2, however, the Monte Carlo evaluation exhibits noticeable inaccuracy at higher z values. This instability arises because the peak of the integrand $t^{z-1}e^{-t}$ shifts dynamically to $t=z-1$ as z increases ($\frac{d}{dt}(t^{z-1}e^{-t}) = 0 \Rightarrow t = z-1$). Our chosen distrubution of $p(t) = e^{-t}$, however, has its peak remaining static at $t=0$. This mismatch causes $g(t)=t^{z-1}$ to amplify massively for high t. Thus, the variance blows up, whic shows that importance sampling fails catastrophically when $p(x)$ no longer tracks the regions of mathematical significance

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.special import gamma

def simpson_third_int(f,a,b,N): #from prev exercise
    if N % 2 != 0:
        print("N must even!")
    else:
        x = np.linspace(a, b, N + 1)
        y = f(x)
        h = (b - a) / N
        return (h / 3) * (y[0] + 4 * np.sum(y[1:-1:2]) + 2 * np.sum(y[2:-2:2]) + y[-1]) #y[1:-1:2] gets the odd indices, while y[2:-2:2] gets the even indices (w3schools ftw)

def J(m,x):
    def integrand(theta):
        return np.cos(m*theta - x*np.sin(theta))
    integral = simpson_third_int(integrand, 0.0, np.pi, 1000)
    return (1 / np.pi) * integral
```

```python
#part a
x_vals = np.linspace(0, 20, 200)

J0_vals = [J(0, x) for x in x_vals]
J1_vals = [J(1, x) for x in x_vals]
J2_vals = [J(2, x) for x in x_vals]

plt.figure(figsize=(12, 6))
plt.plot(x_vals, J0_vals, label='$J_0(x)$', color='red')
plt.plot(x_vals, J1_vals, label='$J_1(x)$', color='green')
plt.plot(x_vals, J2_vals, label='$J_2(x)$', color='blue')

plt.xlabel('x')
plt.ylabel('$J_m(x)$')
plt.axhline(0, color='black', linestyle='--')
plt.legend()
plt.show()
```

```python
#part b
def intensity(r, wavelength=500): #unit: nanometer
    intensity_val = 0
    
    if r == 0:
        intensity_root = 1/2
    else:
        k = 2*np.pi/wavelength
        numerator = J(1, k*r)
        denominator = k*r
        intensity_root = numerator/denominator
        
    return intensity_root**2
```

```python
x = np.linspace(-1000, 1000, 100) #our unit is in nm, and 1um = 1000nm
y = np.linspace(-1000, 1000, 100)

X, Y = np.meshgrid(x, y)
R = np.sqrt(X**2 + Y**2)

I_grid = np.zeros_like(R) #this is where the intensity values will be stored

for i in range(R.shape[0]): #iterates through every row and column in the 2D grid
    for j in range(R.shape[1]):
        I_grid[i, j] = intensity(R[i, j], wavelength=500)

plt.figure(figsize=(10, 10))
plt.imshow(I_grid, origin='lower', extent=[-1, 1, -1, 1], vmax=0.01, cmap='hot')
plt.title(r'Diffraction Pattern of a Point Light Source ($\lambda = 500$ nm)')
plt.xlabel(r'$x$ ($\mu$m)')
plt.ylabel(r'$y$ ($\mu$m)')
plt.savefig("prob1.png")
plt.show()
```

### Problem 2: Integration via Monte Carlo with Importance sampling

Evaluate the gamma function for $z\in [1,5]. $ $$ \Gamma(z) = \int_{0}^{\infty} t^{z-1} e^{-t} \, dt $$ Plot the function using your integration method vs `scipy.special.gamma` (two curves)

Use monte carlo method with importance sampling, by recasting the integral into the form 

$$ \int f(x)dx = \int g(x)p(x)dx, \ \  g(x) = \frac{f(x)}{p(x)}$$


Report discussion: What is the difference between this version of Monte Carlo and the one we did in the previous session? What should we be careful of when applying Monte Carlo with Importance sampling?

(For Simpson's Rule) In order to avoid infinity as the upper bound, we rewrite the function as follows:

![gamma function rewrite](gammafunc.jpg)

```python
def gamma_simpson(z, N=1000):
    def integrand(x):
        x = np.where(x == 1.0, 1.0 - 1e-10, x) #avoid division by zero at x=1
        return (1/(1-x))**2 * (x/(1 - x))**(z-1) * np.exp(-x/(1 - x)) #from the rewrite above to avoid infinity as upper bound
    return simpson_third_int(integrand, 0.0, 1.0, N)
    
def gamma_monte_carlo(z, N=100000): # p(t) = e^(-t), g(t) = t^(z-1)
    t_samples = np.random.exponential(scale=1.0, size=N)
    g_t = t_samples**(z - 1)
    return np.mean(g_t)

z_vals = np.linspace(1, 5, 100)
gamma_simpson_vals = [gamma_simpson(z) for z in z_vals]
gamma_mc_vals = [gamma_monte_carlo(z) for z in z_vals]
gamma_scipy_vals = gamma(z_vals)

plt.figure(figsize=(8, 5))
plt.title("Numerical Approximations of the Gamma Function")
plt.plot(z_vals, gamma_scipy_vals, label='Scipy', color='r', linewidth=3)
plt.plot(z_vals, gamma_simpson_vals, label="Simpson's 1/3 Rule", color='y', linestyle='--')
plt.scatter(z_vals, gamma_mc_vals, label="Monte Carlo", color='g', marker='x')
plt.xlabel('z')
plt.ylabel(r'$\Gamma(z)$')
plt.legend()
plt.savefig("prob2.png")
plt.show()
```
